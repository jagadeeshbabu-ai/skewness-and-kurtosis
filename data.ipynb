```python
import numpy as np
import pandas as pd
from scipy.stats import kurtosis, skew

# 1. Create a sample dataset (let's simulate a right-skewed dataset with heavy tails)
np.random.seed(42)
data = np.concatenate([
    np.random.normal(loc=50, scale=10, size=900),  # Main normal cluster
    np.random.exponential(scale=30, size=100)     # Outliers adding skewness & tails
])

# Convert to a Pandas Series
df_series = pd.Series(data)

# 2. Calculating Skewness
# Using Pandas
pandas_skew = df_series.skew()

# Using SciPy
scipy_skew = skew(data)

print(f"Skewness (Pandas): {pandas_skew:.4f}")
print(f"Skewness (SciPy):  {scipy_skew:.4f}")

# 3. Calculating Kurtosis
# Note: By default, SciPy's kurtosis() calculates 'Fisher's definition' (Excess Kurtosis, where Normal = 0).
# To get the raw Pearson kurtosis (where Normal = 3), use fisher=False.
excess_kurtosis = kurtosis(data, fisher=True)     # Excess Kurtosis (baseline = 0)
pearson_kurtosis = kurtosis(data, fisher=False)   # Pearson Kurtosis (baseline = 3)
pandas_kurtosis = df_series.kurtosis()            # Pandas defaults to Fisher's (Excess Kurtosis)

print(f"Excess Kurtosis (SciPy):  {excess_kurtosis:.4f}")
print(f"Pearson Kurtosis (SciPy): {pearson_kurtosis:.4f}")
print(f"Kurtosis (Pandas):        {pandas_kurtosis:.4f}")
```

```python
import pandas as pd

data = {
    "normal_data": [45, 48, 49, 50, 50, 51, 52, 53, 55],
    "skewed_data": [10, 12, 15, 18, 20, 25, 30, 40, 100]
}

df = pd.DataFrame(data)

print("Normal Data")
print("Skewness:", df["normal_data"].skew())
print("Kurtosis:", df["normal_data"].kurt())

print("\nSkewed Data")
print("Skewness:", df["skewed_data"].skew())
print("Kurtosis:", df["skewed_data"].kurt())
```
